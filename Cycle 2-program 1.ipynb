```python

def min_value(x, y):
    return min(x, y)


drop = 0
count = 0

cap = int(input("Enter the bucket size: "))
process = int(input("Enter the processing rate: "))
nsec = int(input("Enter the number of seconds you want to simulate: "))

inp = []

for i in range(nsec):
    packet = int(input(f"Enter the size of the packet entering at {i + 1} sec: "))
    inp.append(packet)

print("\nSecond | Packet Received | Packet Sent | Packet Left | Dropped")
print("--------------------------------------------------------------")

for i in range(nsec):
    count += inp[i]

    if count > cap:
        drop = count - cap
        count = cap

    print(f"{i + 1}\t {inp[i]}\t\t", end="")

    mini = min_value(count, process)

    print(f"{mini}\t\t", end="")

    count -= mini

    print(f"{count}\t\t {drop}")

    drop = 0

# Process remaining packets
i = nsec

while count != 0:
    if count > cap:
        drop = count - cap
        count = cap

    print(f"{i + 1}\t 0\t\t", end="")

    mini = min_value(count, process)

    print(f"{mini}\t\t", end="")

    count -= mini

    print(f"{count}\t\t {drop}")

    drop = 0
    i += 1
```
